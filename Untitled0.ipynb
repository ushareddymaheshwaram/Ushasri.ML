```python
import os
import numpy as np

print(os.listdir('/content/dataset'))
```

```python
import pandas as pd

df = pd.read_csv('/content/dataset/adult.csv')

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
# Display all column names
print("Column names:")
print(df.columns.tolist())

# Display first 5 rows
df.head()
```

```python
# Check data types and non-null values
df.info()

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check for '?' values
print("\nValues represented by '?':")
print((df == '?').sum())
```

```python
# Replace '?' with NaN
df = df.replace('?', np.nan)

# Check missing values again
print("Missing values after replacing '?':")
print(df.isnull().sum())
```

```python
# Find categorical columns
categorical_columns = df.select_dtypes(include='object').columns

# Fill missing categorical values with the mode
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Check that missing values have been handled
print("Remaining missing values:")
print(df.isnull().sum())
```

```python
# Count duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Remove duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# Display summary statistics for numerical columns
df.describe().T
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Boxplots for numerical variables
numeric_columns = [
    'age',
    'fnlwgt',
    'educational-num',
    'capital-gain',
    'capital-loss',
    'hours-per-week'
]

plt.figure(figsize=(12, 6))

df[numeric_columns].boxplot()

plt.title("Boxplots of Numerical Variables")
plt.xticks(rotation=45)
plt.ylabel("Value")
plt.show()
```

```python
# Cap extreme fnlwgt values using the 1st and 99th percentiles
lower = df['fnlwgt'].quantile(0.01)
upper = df['fnlwgt'].quantile(0.99)

df['fnlwgt'] = df['fnlwgt'].clip(lower, upper)

print("fnlwgt values capped between:")
print(lower, "and", upper)
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='age', bins=30, kde=True)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Number of People")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='income')

plt.title("Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of People")

plt.show()
```

```python
plt.figure(figsize=(10, 7))

# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Convert income into a binary variable for analysis
df['income_binary'] = (df['income'] == '>50K').astype(int)

education_income = (
    df.groupby('education')['income_binary']
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    x=education_income.values,
    y=education_income.index
)

plt.title("Proportion Earning More Than $50K by Education")
plt.xlabel("Proportion Above $50K")
plt.ylabel("Education")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df,
    x='income',
    y='hours-per-week'
)

plt.title("Working Hours by Income Group")
plt.xlabel("Income")
plt.ylabel("Hours per Week")

plt.show()
```

```python
# Feature 1: Net capital
df['net_capital'] = df['capital-gain'] - df['capital-loss']

print(df[['capital-gain', 'capital-loss', 'net_capital']].head())
```

```python
# Feature 2: Long working hours indicator
df['long_hours'] = (df['hours-per-week'] >= 50).astype(int)

print(df[['hours-per-week', 'long_hours']].head(10))
```

```python
# Feature 3: Age group
df['age_group'] = pd.cut(
    df['age'],
    bins=[16, 25, 35, 45, 55, 65, 100],
    labels=['17-25', '26-35', '36-45', '46-55', '56-65', '66+']
)

print(df[['age', 'age_group']].head(10))
```

```python
# Feature 4: Married indicator
df['is_married'] = (
    df['marital-status']
    .str.contains('Married', case=False, regex=False)
    .astype(int)
)

print(df[['marital-status', 'is_married']].head(10))
```

```python
print("Final dataset shape:", df.shape)

print("\nTotal missing values:")
print(df.isnull().sum().sum())

print("\nTotal duplicate rows:")
print(df.duplicated().sum())

print("\nNew features created:")
print([
    'net_capital',
    'long_hours',
    'age_group',
    'is_married'
])
```

# 5 Key Insights

1. **Income distribution is imbalanced:** Most individuals in the dataset earn <=50K, while a smaller proportion earn >50K. This should be considered when developing a future classification model.

2. **Education is associated with income:** Higher educational qualifications generally correspond to a higher proportion of individuals earning more than $50K.

3. **Working hours show a relationship with income:** Individuals earning more than $50K tend to have higher working hours compared with the lower-income group.

4. **Age is an important demographic factor:** Income patterns vary across age groups, with middle-aged individuals generally showing higher proportions of income above $50K than younger groups.

5. **Data preprocessing was necessary:** Missing categorical values represented by '?' were converted to missing values and filled using the mode. Duplicate records were removed, and extreme values in `fnlwgt` were capped to reduce their influence.
